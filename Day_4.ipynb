#Sets

Are built in data type used to store collection of unique items, They are unordered and do not allow duplicate entries, sets are identical while performing mathemetical set operations like union, intersection, difference ...


```python
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
emp_set = set()
print(type(emp_set))
```

```python
emp_set = set([1,2,3,4,5,6])
print(emp_set)
print(type(emp_set))
```

```python
#Sets do not allow duplicates

my_set = set([1,2,3,4,5,6,4,5,6])
print(my_set)
```

```python
#Basic sets operation
#Add and Remove elements

my_set.add(7)
print(my_set)

```

```python
my_set.remove(3)
print(my_set)
```

```python
#Discard will remove element from the cell if it is a member

my_set.discard(3)
print(my_set)
```

```python
#POP method

a = {1,2,4,5,6,7,8,9,10}
removed_element = a.pop()
print(removed_element)
print(a)

#As this is first in first out, First element will be removed
```

```python
#clear all elements

a.clear()
print(a)
```

```python
#Set Membership

new_set = {23,45,67,89,90}
print(23 in new_set)
print(34 in new_set)
```

```python
#Mathematical Operations

set1 = {1,2,3,4,5,6}
set2 = {3,4,5,6,7,8}

#Union
Union_set = set1.union(set2)
print(Union_set)

#Intersection
int_set = set1.intersection(set2)
print(int_set)
```

```python

set1 = {3,4,5,6,7,8,9}
set2 = {5,6,7,8,9,10,11}

#Difference - Commom elements will be removed

print(set1.difference(set2))
```

```python
print(set1.difference_update(set2))

#Updates the output in set1
```

```python
print(set2.difference(set1))
```

```python
#Symmetric Difference - Intersection elements will be removed

print(set1.symmetric_difference(set2))
```
