#Dictionaries

##Dictionaries -  Unordered collection of items must stored in key-value pairs, key must be unique and values can be of any type

```python
 #create dict
 emp_dict = {}
 print(type (emp_dict))
```

```python
employee={'name':'Anirudh',"age":26, 'empid':3211}
print(employee)
```

```python
#Accessing dict elements
print(employee['age'])
print(employee['name'])

#Accessing using get method
print(employee.get('empid'))
print(employee.get('age'))

print(employee.get('last_name','Nayak'))    #defaut value is replaced if not present
```

#Modifying Dict Elements
##Dict are mutable, so you can add, update or delete elements



```python
print(employee)
```

```python
employee['age']=27  #Update key
```

```python
#Add key
print(employee)
employee['last_name']='nayak'
```

```python
#delete a key

del employee['last_name']
print(employee)
```

#Dictionary methods

```python

keys = employee.keys()
print(keys)

values=employee.values()
print(values)

items=employee.items()
print(items)
```

```python
##Shallow copy

employee_copy = employee
print(employee_copy)
```

```python
employee['name']="Andy"
print(employee)
print(employee_copy)    #same update will be copied even if you don't want to copy and need different file
```

```python
employee_copy1=employee.copy()
print(employee_copy1)
print(employee)
```

```python
employee['name']="Anirudh_nayak"
print(employee_copy1)
print(employee)       #Shallow copy provides different memory locations, whenever we change one will not impact another
```

#Iterating over Dictionaries

```python
#can iterate over dict, keys, values or items
```

```python
#Iterating over keys, values
for keys in employee:
    print(keys)
```

```python
#Iterating over values

for values in employee.values():
    print(values)
```

```python
#Iterate over key value pair

for key,value in employee.items():
    print(key,value)
```

```python
#Nested Dictionaries
#Nested Dictionaries

employee={
    "employee1":{"name":"Anni","age":27},
    "employee2":{"name":"Nayak","age":30}

}
print(employee)
```

```python
#Accessing nested dict

print(employee["employee1"]["age"])
```

##Dictionary Comprahensions

```python
square = {x:x**2 for x in range(5)}
print(square)
```

```python
#conditional dict comp

evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
#count frequency of numbers(each elements) in list

numbers = [1,2,2,3,3,3,4,4,4,4]
freq = {}

for num in numbers:
  if num in freq:
    freq[num]+=1
  else:
    freq[num]=1
print(freq)

```

```python
#Merge 2 dict into 1

dict1={'a':1,'b':2}
dict2={'b':3,'c':4}
merged_dict={**dict1,**dict2}     #Using ** - Any number of key,value pair is taken and appended to that new dict
print(merged_dict)
```
